## 🧠 Understanding Basic Concepts

Before working with data, it is important to understand a few key terms.

---

### 📄 What is a CSV file?

CSV stands for **Comma-Separated Values**.

It is a simple file format used to store tabular data.

Example:
```
user_id,event_type,location
101,login,India
102,purchase,US
```

Each row represents a record, and each column represents a field.

---

### 📊 What is a DataFrame?

A **DataFrame** is a table-like structure in Python (provided by pandas).

It looks like an Excel sheet:

| user_id | event_type | location |
|--------|-----------|----------|
| 101 | login | India |
| 102 | purchase | US |

---

### 🧠 Why DataFrame is Important?

It allows you to:
- Filter data  
- Find missing values  
- Detect duplicates  
- Analyze patterns  

---

### 🧰 What is pandas?

`pandas` is a Python library used for:
- Reading data (CSV, Excel, etc.)
- Cleaning data  
- Analyzing data  

---

### 🔁 Real-world Flow
```
CSV File → pandas → DataFrame → Analysis → Debugging
```
